# Pandas Basics: Loading and Exploring a Student Score Dataset

### Objective
In this notebook, we will:
- Import the Pandas library.
- Load a Student Score dataset.
- Explore the dataset.
- View rows and columns.
- Display dataset information.
- Check for missing values.
- Generate summary statistics.

**Expected Outcome:** The dataset is successfully loaded and explored using Pandas.

```python
# Import Pandas
import pandas as pd
```

## Step 1: Load the Dataset

Load the Student Score dataset using `read_csv()`.

> Ensure the dataset file is present in the same directory as this notebook.

```python
# Load dataset
df = pd.read_csv("/content/student_scores.csv")

# Display first five rows
df.head()
```

## Step 2: Display the Last Five Rows

The `tail()` function displays the last five records of the dataset.

```python
df.tail()
```

## Step 3: Find Dataset Dimensions

The `shape` attribute returns the number of rows and columns.

```python
print("Dataset Shape:", df.shape)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

## Step 4: Display Column Names

The `columns` attribute lists all column names in the dataset.

```python
print(df.columns)
```

## Step 5: Display Dataset Information

The `info()` function provides:
- Number of rows
- Number of columns
- Data types
- Missing values
- Memory usage

```python
df.info()
```

## Step 6: Generate Summary Statistics

The `describe()` function returns statistical information for numerical columns.

```python
df.describe()
```

## Step 7: Check Missing Values

Use `isnull().sum()` to count missing values in each column.

```python
df.isnull().sum()
```

## Step 8: Display Dataset Index

The `index` attribute shows the row labels of the dataset.

```python
df.index
```

## Step 9: Display Data Types

The `dtypes` attribute returns the data type of each column.

```python
df.dtypes
```
