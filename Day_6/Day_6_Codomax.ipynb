# Data Visualization with Matplotlib

### Objective
In this notebook, we will:

- Load the Student Score dataset.
- Create a Scatter Plot.
- Create a Bar Chart.
- Create a Line Chart.

**Expected Outcome:** Basic charts are successfully created using Matplotlib.

```python
# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
```

## Step 1: Load the Dataset

Load the student score dataset.

```python
df = pd.read_csv("/content/student_scores.csv")

df.head()
```

## Step 2: Scatter Plot

Scatter plots help visualize the relationship between two numerical variables.

We will plot:
- Hours Studied (X-axis)
- Scores (Y-axis)

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Hours_Studied"], df["Scores"])

plt.title("Hours Studied vs Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Scores")

plt.show()
```

## Step 3: Bar Chart

Bar charts are useful for comparing values across categories.

We will display the scores of the first 10 students.

```python
top10 = df.head(10)

plt.figure(figsize=(10,5))

plt.bar(top10["Name"], top10["Scores"])

plt.title("Scores of First 10 Students")
plt.xlabel("Students")
plt.ylabel("Scores")

plt.xticks(rotation=45)

plt.show()
```

## Step 4: Line Chart

Line charts help show trends over a sequence.

We will plot Student ID against Scores.

```python
plt.figure(figsize=(8,5))

plt.plot(df["Student_ID"], df["Scores"])

plt.title("Student ID vs Scores")
plt.xlabel("Student ID")
plt.ylabel("Scores")

plt.show()
```

## Step 5: Save Charts (Optional)

Save the generated charts as image files.

```python
# Scatter Plot
plt.figure(figsize=(8,5))
plt.scatter(df["Hours_Studied"], df["Scores"])
plt.title("Hours Studied vs Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Scores")
plt.savefig("scatter_plot.png")
plt.close()

# Bar Chart
plt.figure(figsize=(10,5))
plt.bar(top10["Name"], top10["Scores"])
plt.title("Scores of First 10 Students")
plt.xlabel("Students")
plt.ylabel("Scores")
plt.xticks(rotation=45)
plt.savefig("bar_chart.png")
plt.close()

# Line Chart
plt.figure(figsize=(8,5))
plt.plot(df["Student_ID"], df["Scores"])
plt.title("Student ID vs Scores")
plt.xlabel("Student ID")
plt.ylabel("Scores")
plt.savefig("line_chart.png")
plt.close()

print("Charts saved successfully!")
```
