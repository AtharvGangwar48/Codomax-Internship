# Day 11: Prediction App

## Objective
Create a simple interactive Python program within this notebook where users can enter their study hours and receive a predicted exam score based on our trained Linear Regression model.

### Key Tasks:
1. Load the student score dataset.
2. Train the Linear Regression model.
3. Create an interactive input cell to prompt the user for input.
4. Validate the input (check if it is a positive number).
5. Generate and display the predicted score (capped between 0% and 100%).

**Expected Outcome:** A functional interactive cell that predicts student scores based on user input.

## Step 1: Import Libraries and Train Model
First, we quickly load the data and train the Linear Regression model on all available student score records to maximize prediction accuracy.

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

# Load data
try:
    df = pd.read_csv("student_scores.csv")
except FileNotFoundError:
    df = pd.read_csv("/content/student_scores.csv")

# Train model on the entire dataset to get the best fit coefficients
X = df[["Hours_Studied"]]
y = df["Scores"]

model = LinearRegression()
model.fit(X, y)

print("Model trained successfully and ready for predictions!")
print(f"Equation: Score = {model.coef_[0]:.4f} * Hours_Studied + {model.intercept_:.4f}")
```

## Step 2: Interactive Prediction Cell
Run the cell below, enter your study hours when prompted, and press Enter to see your predicted score.

```python
def predict_user_score():
    print("==================================================")
    print("      STUDENT SCORE PREDICTOR APP (Day 11)        ")
    print("==================================================")
    
    user_input = input("Enter study hours per day (e.g., 5.5): ")
    
    try:
        hours = float(user_input)
        if hours < 0:
            print("❌ Error: Study hours cannot be negative. Please enter a valid number.")
            return
        elif hours > 24:
            print("⚠️ Warning: A day only has 24 hours! Setting hours to 24.")
            hours = 24.0
            
        # Predict using model
        predicted_arr = model.predict([[hours]])
        predicted_score = predicted_arr[0]
        
        # Cap predicted score between 0 and 100
        final_score = max(0.0, min(100.0, predicted_score))
        
        print("\n--- Results ---")
        print(f"Study Hours     : {hours:.2f} hours")
        print(f"Raw Prediction  : {predicted_score:.2f} marks")
        print(f"Final Score Cap : {final_score:.2f}%")
        print("--------------------------------------------------")
        
        # Provide a friendly qualitative feedback
        if final_score >= 90:
            print("🎉 Excellent! Excellent score prediction (Grade A).")
        elif final_score >= 75:
            print("👍 Very Good! Keep studying to maintain this grade.")
        elif final_score >= 50:
            print("📚 Good! Regular study will improve your scores further.")
        else:
            print("💡 Needs improvement! Focus on increasing daily study hours.")
            
    except ValueError:
        print("❌ Error: Invalid input. Please enter a valid numerical value.")

# Run the interactive function
predict_user_score()
```
