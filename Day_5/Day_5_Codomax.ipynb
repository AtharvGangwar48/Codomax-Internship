# Data Cleaning Using Pandas

### Objective

In this notebook, we will:
- Load the Student Score dataset.
- Identify missing values.
- Handle missing values.
- Detect duplicate records.
- Remove duplicates.
- Generate descriptive statistics.

**Expected Outcome:** A clean dataset is prepared for further analysis.

```python
# Import Pandas
import pandas as pd
```

## Step 1: Load the Dataset

Load the CSV file into a Pandas DataFrame.

```python
# Load dataset
df = pd.read_csv("/content/student_scores_dirty.csv")

# Display first five rows
df.head()
```

## Step 2: Check Missing Values

The `isnull().sum()` function counts missing values in each column.

```python
print("Missing Values:")
df.isnull().sum()
```

## Step 3: Handle Missing Values

Fill missing numerical values with the column mean.

> If the dataset contains no missing values, the dataset remains unchanged.

```python
numeric_columns = df.select_dtypes(include=["number"]).columns

df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].mean()
)

df.head()
```

## Step 4: Verify Missing Values Again

Confirm that all missing values have been handled.

```python
df.isnull().sum()
```

## Step 5: Check for Duplicate Records

The `duplicated()` function identifies duplicate rows.

```python
print("Number of Duplicate Rows:")
print(df.duplicated().sum())
```

## Step 6: Remove Duplicate Records

Remove duplicate rows using `drop_duplicates()`.

```python
df = df.drop_duplicates()

print("Dataset Shape After Removing Duplicates:")
print(df.shape)
```

## Step 7: Display Dataset Statistics

Generate descriptive statistics for numerical columns.

```python
df.describe()
```

## Step 8: Display Dataset Information

Check the structure and data types after cleaning.

```python
df.info()
```

## Step 9: Save the Cleaned Dataset

Store the cleaned dataset as a new CSV file.

```python
df.to_csv("student_scores_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

# Conclusion

In this notebook, we successfully:

- Loaded the dataset.
- Identified missing values.
- Filled missing numerical values.
- Checked for duplicate records.
- Removed duplicate rows.
- Analyzed dataset statistics.
- Saved the cleaned dataset.

**Expected Outcome Achieved:** A clean dataset is prepared for further analysis.
